## Streamlit | Structure de base

- Chaque script Streamlit commence par `import streamlit as st`
- Les fonctions s'affichent de haut en bas (aucun layout complexe par défaut)
- Chaque modification du code rafraichit automatiquement l'application
- Streamlit supporte le hot reloading

## Streamlit | Comment ça marche? (Le modèle d'exécution)

**Streamlit fonctionne comme un script Python classique qui se ré-exécute de haut en bas à chaque interaction utilisateur.**

Quand vous interagissez avec un widget (bouton, slider, etc.):
1. Le navigateur envoie l'information au serveur Streamlit
2. Streamlit ré-exécute **tout le script** depuis le début
3. Tous les widgets sont recréés avec leurs nouvelles valeurs
4. L'interface se met à jour avec les nouveaux résultats

C'est pourquoi on peut utiliser `if st.button(...)` - le widget retourne `True` quand on clique, et tout le code est ré-exécuté!

## Streamlit | Pourquoi `if widget` fonctionne?

**Chaque widget Streamlit retourne une valeur** qui peut être stockée dans une variable:

| Widget | Retourne |
|--------|----------|
| `st.button()` | `True` si cliqué, `False` sinon |
| `st.checkbox()` | `True` si coché, `False` sinon |
| `st.radio()` | L'option sélectionnée (string) |
| `st.selectbox()` | L'option sélectionnée (string) |
| `st.slider()` | La valeur sélectionnée (int/float) |
| `st.text_input()` | Le texte saisi (string) |
| `st.number_input()` | Le nombre saisi (int/float) |

```python
# Le bouton retourne True quand on clique
if st.button("Cliquez-moi"):
    st.write("Bouton cliqué!")

# Le slider retourne la valeur actuelle
age = st.slider("Âge", 0, 100, 25)
st.write(f"Vous avez {age} ans")
```

## Streamlit | Exemple: graphique qui se met à jour

Voici l'exemple le plus important - un graphique qui change quand on modifie le slider:

```python
import streamlit as st
import matplotlib.pyplot as plt
import numpy as np

st.title("Graphique interactif")

# 1. Le widget crée une variable qui change à chaque interaction
amplitude = st.slider("Amplitude", 0.1, 5.0, 1.0)

# 2. On utilise cette variable pour générer les données
x = np.linspace(0, 10, 100)
y = amplitude * np.sin(x)

# 3. Le graphique est recréé à chaque modification du slider
fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_ylim(-5, 5)
st.pyplot(fig)
```

**Ce qui se passe:**
1. L'utilisateur bouge le curseur
2. Streamlit détecte le changement → "rerun"
3. Le script se ré-exécute avec la nouvelle valeur de `amplitude`
4. Le graphique est recréé avec la nouvelle amplitude

## Streamlit | Cycle de vie d'une application

```
┌─────────────────────────────────────────────────────┐
│  1. Utilisateur ouvre l'app                          │
│     → Script exécuté de haut en bas                 │
│     → Tous les widgets créés                       │
│     → Interface affichée                            │
└─────────────────────────────────────────────────────┘
                          ↓
┌─────────────────────────────────────────────────────┐
│  2. Utilisateur interagit (clic, slider, etc.)      │
│     → Streamlit détecte le changement                │
│     → Envoi d'un signal au serveur                  │
└─────────────────────────────────────────────────────┘
                          ↓
┌─────────────────────────────────────────────────────┐
│  3. Script rerun (ré-exécution complète)            │
│     → Widgets recréés avec nouvelles valeurs       │
│     → Variables contiennent les nouvelles valeurs  │
│     → Graphiques/tableaux mis à jour               │
└─────────────────────────────────────────────────────┘
                          ↓
┌─────────────────────────────────────────────────────┐
│  4. Interface mise à jour                            │
│     → Retour à l'état d'attente                    │
└─────────────────────────────────────────────────────┘
```

## Streamlit | Ordre d'exécution - IMPORTANT

**L'ordre du code compte!** Tout s'exécute de haut en bas:

```python
import streamlit as st

st.title("Mon titre")          # 1. Titre en haut
if st.button("Bouton"):         # 2. Bouton au milieu
    st.write("Message")         # 3. Message en bas (apparaît si clic)
```

ATTENTION: Ne vous méprenez pas sur l'ordre:
- Le `if` ne contrôle PAS où apparaît le message
- Le message apparaît toujours après le bouton dans le flux
- Le `if` contrôle seulement SI le message apparaît

## Streamlit | st.button() et st.checkbox()

`st.button()`: Crée un bouton cliquable
`st.checkbox()`: Crée une case à cocher

**Rappel:** Chaque exécution du script recrée tous les widgets. Si le bouton est cliqué, il retourne `True` pendant cette exécution.

````python
import streamlit as st

# Compteur qui s'incrémente à chaque clic
if 'clicks' not in st.session_state:
    st.session_state.clicks = 0

def increment():
    st.session_state.clicks += 1

# Bouton avec callback (fonction appelée au clic)
st.button('Cliquez-moi', on_click=increment)
st.write(f'Nombre de clics: {st.session_state.clicks}')
````

````python
# Vérification simple avec if
if st.button('Test'):
    st.write('Vous avez cliqué!')
else:
    st.write('En attente de clic...')
````

## Streamlit | Affichage conditionnel de données

On peut utiliser les widgets pour contrôler l'affichage de graphiques, tableaux, etc.

````python
import streamlit as st
import pandas as pd
import numpy as np

st.title("Affichage conditionnel")

# Checkbox pour afficher/masquer les données
show_data = st.checkbox("Afficher les données détaillées")

# Checkbox pour afficher/masquer le graphique
show_chart = st.checkbox("Afficher le graphique")

# Générer des données
df = pd.DataFrame({
    'x': range(50),
    'y': np.random.randn(50).cumsum()
})

# Afficher selon l'état des checkboxes
if show_data:
    st.subheader("Données")
    st.dataframe(df)

if show_chart:
    st.subheader("Graphique")
    st.line_chart(df.set_index('x')['y'])

st.write("---")
st.write(f"Données affichées: {show_data}, Graphique affiché: {show_chart}")
````

## Streamlit | Comment le graphique se met à jour?

Reprenons l'exemple précédent:

```python
if show_chart:  # show_chart est True ou False
    st.subheader("Graphique")
    st.line_chart(df.set_index('x')['y'])
```

**Explication:**

1. L'utilisateur coche la case "Afficher le graphique"
2. `st.checkbox()` retourne `True`
3. Streamlit détecte le changement → rerun
4. Le script s'exécute à nouveau
5. `show_chart = True` maintenant
6. Le `if show_chart:` est évalué à `True`
7. Le graphique est affiché

Si on decoche, `show_chart = False`, le graphique disparaît!

## Streamlit | Graphiques avec Matplotlib

Les graphiques matplotlib se mettent à jour automatiquement quand leurs données changent!

````python
import streamlit as st
import matplotlib.pyplot as plt
import numpy as np

st.write("**Modifiez les valeurs ci-dessous et observez le graphique se mettre à jour!**")

# Ces variables prennent les valeurs des widgets
amplitude = st.slider('Amplitude du sinus', 0.1, 3.0, 1.0, 0.1)
frequence = st.slider('Fréquence', 1, 10, 1)
phase = st.slider('Phase (degrés)', 0, 360, 0)

# Génération des données basée sur les valeurs des widgets
x = np.linspace(0, 4 * np.pi, 200)
y = amplitude * np.sin(frequence * x + np.radians(phase))

# Création et affichage du graphique
fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(x, y, 'b-', linewidth=2)
ax.axhline(y=0, color='k', linewidth=0.5)
ax.set_xlim(0, 4 * np.pi)
ax.set_ylim(-4, 4)
ax.set_xlabel('x')
ax.set_ylabel('sin(x)')
ax.set_title(f'sin(x) avec amplitude={amplitude}, fréquence={frequence}, phase={phase}°')
ax.grid(True, alpha=0.3)

st.pyplot(fig)
````

## Streamlit | Session State - Persistance des données

**Problème:** Quand Streamlit ré-exécute le script, les variables sont réinitialisées.

**Solution:** `st.session_state` permet de persister des données entre les exécutions.

```python
# Sans session_state - le compteur revient à 0 à chaque exécution
compteur = 0  # Sempre réinitialisé!
if st.button("Incrémenter"):
    compteur += 1
st.write(compteur)

# Avec session_state - la valeur persiste
if 'compteur' not in st.session_state:
    st.session_state.compteur = 0  # Initialisation
if st.button("Incrémenter"):
    st.session_state.compteur += 1  # Persiste entre les exécutions
st.write(st.session_state.compteur)
```

````python
import streamlit as st

# Initialiser le compteur
if 'counter' not in st.session_state:
    st.session_state.counter = 0

# Fonctions pour incrémenter/décrémenter
def increment():
    st.session_state.counter += 1

def decrement():
    st.session_state.counter -= 1

# Afficher et manipuler le compteur
st.write(f'Compteur: {st.session_state.counter}')
st.button('Incrémenter', on_click=increment)
st.button('Décrémenter', on_click=decrement)
````

## Exemple complet: Application d'analyse de données

Cet exemple combine plusieurs concepts expliqués précédemment.

````python
import streamlit as st
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px

st.title("Application d'Analyse de Données")

# ============================================
# SECTION 1: Sidebar pour les contrôles
# ============================================
st.sidebar.title("Paramètres")
uploaded_file = st.sidebar.file_uploader("Choisissez un fichier CSV", type=['csv'])

# ============================================
# SECTION 2: Contenu principal
# ============================================
if uploaded_file is not None:
    # Lire les données
    df = pd.read_csv(uploaded_file)
    
    # Afficher les données
    st.header("Données")
    st.dataframe(df.head(100))
    
    # Statistiques avec colonnes
    st.header("Statistiques")
    col1, col2, col3 = st.columns(3)
    col1.metric("Lignes", df.shape[0])
    col2.metric("Colonnes", df.shape[1])
    col3.metric("Valeurs manquantes", df.isnull().sum().sum())
    
    # Sélection de colonne pour le graphique
    numeric_cols = df.select_dtypes(include='number').columns.tolist()
    if numeric_cols:
        # Le selectbox retourne la colonne choisie
        col_select = st.selectbox("Sélectionnez une colonne numérique:", numeric_cols)
        
        # Le slider contrôle le nombre de bins
        bins = st.slider("Nombre de bins", 5, 100, 30)
        
        # Le graphique se met à jour automatiquement
        st.header(f"Distribution de {col_select}")
        
        fig, ax = plt.subplots()
        ax.hist(df[col_select].dropna(), bins=bins, edgecolor='black')
        ax.set_xlabel(col_select)
        ax.set_ylabel("Fréquence")
        st.pyplot(fig)
else:
    st.info("Téléchargez un fichier CSV pour commencer.")

# ============================================
# L'application se ré-exécute entièrement à chaque modification
# ============================================
````

## Streamlit | Configuration de page

`st.set_page_config()` permet de configurer la page (titre, icône, layout).

````python
import streamlit as st

# Configuration de la page (DOIT ÊTRE EN PREMIER)
st.set_page_config(
    page_title='Mon Application',
    page_icon='📊',
    layout='wide',
    initial_sidebar_state='expanded'
)

st.title('Application Configurée')
st.write('La page est configurée avec un titre, une icône et un layout large.')
````

## Streamlit | Déploiement

Pour déployer une application Streamlit:

1. **Streamlit Cloud** (recommandé):
   - Créer un compte sur share.streamlit.io
   - Connecter votre repository GitHub
   - Déployer en quelques clics

2. **Heroku**:
   - Créer un fichier `requirements.txt`
   - Créer un fichier `Procfile`
   - Déployer avec git

3. **Docker**:
   - Créer un Dockerfile
   - Builder et déployer le conteneur

## Streamlit | Bonnes pratiques

- **Organisation du code**: Structurez votre application en sections claires
- **Performance**: Utilisez `@st.cache` pour les fonctions coûteuses
- **UX**: Utilisez la sidebar pour les contrôles et le contenu principal pour les résultats
- **Données**: Validez les entrées utilisateur avant de les utiliser
- **Session State**: Persistez l'état quand nécessaire
- **Tests**: Testez votre application régulièrement
