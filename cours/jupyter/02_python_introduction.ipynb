# Qu'est-ce que Python ?

**Python** est un langage de programmation :
- **de haut niveau** – proche du langage naturel, loin de la machine,
- **interprété** – exécuté par un programme tiers (l'interpréteur),
- **à usage général** – utilisable pour le web, la data science, l'IA, les scripts, l'embarqué, etc.,
- **multi-paradigme** – on peut coder en impératif, orienté objet ou fonctionnel.

Sa grande force : une syntaxe **lisible** et **concise**, qui se rapproche du pseudo-code.

# Un peu d'histoire

Python a été conçu par **Guido van Rossum** aux Pays-Bas à la fin des années 1980.

- **Décembre 1989** – Guido commence le projet pendant les vacances de Noël au CWI (Amsterdam).
- **Février 1991** – publication de la version **0.9.0** sur `alt.sources`, première version publique.
- **2000** – **Python 2.0** : Unicode, list comprehensions, ramasse-miettes (garbage collector).
- **2008** – **Python 3.0** : refonte non rétro-compatible du langage.
- **2020** – fin officielle du support de Python 2.
- **Aujourd'hui** – Python 3.12 / 3.13, devenu l'un des langages les plus utilisés au monde.

> Le nom **Python** ne vient pas du serpent, mais du groupe d'humoristes britanniques *Monty Python's Flying Circus* dont Guido était fan. Beaucoup d'exemples de la documentation officielle y font référence.

# Langage compilé vs langage interprété

Pour exécuter un programme, l'ordinateur ne comprend que des instructions binaires. Il faut donc traduire le code source. Deux grandes familles :

| Langage compilé (C, C++, Go, Rust) | Langage interprété (Python, JavaScript, Ruby) |
|-------------------------------------|-----------------------------------------------|
| Un **compilateur** traduit tout le code en un exécutable (binaire) | Un **interpréteur** lit et exécute directement le code source |
| Plus rapide à l'exécution | Plus lent, mais plus flexible et interactif |
| Erreurs détectées à la compilation | Erreurs détectées pendant l'exécution |
| Distribution facile d'un binaire | Il faut l'interpréteur pour exécuter le script |

> En pratique, Python est **semi-interprété** : le source est d'abord traduit en *bytecode* (fichiers `.pyc`), puis ce bytecode est exécuté par la **Machine Virtuelle Python** (PVM).

# L'interpréteur Python

L'**interpréteur** est le programme qui lit votre code et l'exécute. On l'invoque généralement avec la commande `python` ou `python3`.

Il s'utilise principalement de deux manières :

### 1. Mode interactif (REPL)
*Read – Eval – Print – Loop* : on tape une instruction, l'interpréteur l'exécute immédiatement et affiche le résultat, puis on recommence. Pratique pour explorer.

### 2. Mode script
On enregistre un ensemble d'instructions dans un fichier texte `.py`, puis on demande à l'interpréteur d'exécuter ce fichier d'un coup. Pratique pour des programmes complets et réutilisables.

# Le fichier `.py` – un programme

Un programme Python est un simple **fichier texte** avec l'extension `.py`. Il contient une suite d'instructions lues séquentiellement par l'interpréteur.

Exemple – fichier `hello.py` :
```python
# hello.py
print("Hello, World!")
print("Le résultat de 2 + 3 est", 2 + 3)
```

Pour l'exécuter depuis le terminal :
```bash
$ python3 hello.py
Hello, World!
Le résultat de 2 + 3 est 5
```

Un `.py` peut s'éditer avec n'importe quel éditeur de texte : VS Code, PyCharm, Vim, Nano, Sublime Text, etc.

# Qu'est-ce qu'une instruction ?

Une **instruction** (en anglais *statement*) est une **unité d'exécution** que l'interpréteur peut comprendre et exécuter. Une instruction demande à l'ordinateur de faire quelque chose.

Exemples d'instructions :
```python
print("Bonjour")          # un appel de fonction
x = 10                     # une assignation
import math                # un import de module
if x > 0: print("positif") # un test conditionnel
```

Une instruction peut aussi produire une **expression** évaluée :
```python
2 + 3 * 4                  # expression évaluée à 14
```

# La ligne de code en Python

Contrairement à des langages comme C, Java ou JavaScript, **Python n'utilise pas de point-virgule `;`** pour terminer une instruction.

La **fin de ligne** marque la fin d'une instruction. C'est le retour à la ligne qui sert de séparateur.

```python
print("ligne 1")
x = 10
y = x + 5
```

Deux exceptions :
1. On peut mettre plusieurs instructions sur une même ligne en les séparant par `;` (mais **ce n'est pas recommandé**).
2. Une instruction peut s'étaler sur plusieurs lignes :
   - avec un backslash `\` à la fin de la ligne,
   - ou plus proprement, avec des parenthèses `()`, crochets `[]` ou accolades `{}`.

````python
print("une instruction par ligne")
x = 10
y = x + 5
print(y)
````

````python
print("a"); print("b"); print("c – tout sur une ligne, à éviter")
````

````python
total = (1 + 2
         + 3 + 4
         + 5)
print("total =", total)
````

````python
somme = 1 + 2 + 3 \
      + 4 + 5 + 6
print("somme =", somme)
````

# Les commentaires

Un **commentaire** est une portion de texte ignorée par l'interpréteur. Il est destiné aux humains qui lisent le code.

```python
# Tout le reste de cette ligne est un commentaire
x = 10   # commentaire en bout de ligne
```

Pour un commentaire sur plusieurs lignes, on répète le `#` sur chaque ligne.

On peut aussi utiliser une **docstring** (chaîne entre triples guillemets `"""`) comme commentaire multiligne, surtout pour documenter une fonction ou un module.

````python
# Cette ligne est un commentaire – ignorée par l'interpréteur
print("Ceci est exécuté")   # commentaire en bout de ligne

"""
Ceci est une docstring.
Techniquement une chaîne de caractères non assignée,
donc ignorée à l'exécution. Utilisée pour documenter.
"""
print("Fin")
````

# L'indentation, c'est important !

En Python, **l'indentation** (les espaces en début de ligne) n'est pas qu'esthétique : elle **définit les blocs de code**. C'est ce qui remplace les accolades `{}` qu'on trouve en C, Java ou JavaScript.

```python
if True:
    print("dans le bloc if")       # indenté – fait partie du if
    print("encore dans le bloc")   # indenté – fait partie du if
print("en dehors du bloc")         # non indenté – dans le programme principal
```

Une erreur d'indentation provoque une `IndentationError`.

**Convention PEP 8** : 4 espaces par niveau d'indentation. **Il ne faut pas mélanger des espaces et des tabulations**.

````python
if True:
    print("niveau 1")
    if True:
        print("niveau 2")
        if True:
            print("niveau 3")
    print("retour niveau 1")
print("programme principal")
````

# Les bases de la programmation

Ce cours s'adresse à des personnes qui **n'ont jamais programmé**. Avant d'aller plus loin dans la syntaxe Python, prenons un moment pour comprendre ce que signifie *programmer*, et adopter les bons réflexes dès le début.

Programmer, c'est **donner à un ordinateur une suite d'instructions** pour qu'il réalise une tâche à notre place. L'ordinateur ne devine pas : il fait **exactement** ce qu'on lui dit — ni plus, ni moins.

# Un programme, c'est quoi ?

Un **programme** (ou *script*) est une suite d'instructions exécutées une à une, dans l'ordre, par l'ordinateur.

On peut le voir comme une **recette de cuisine** :
1. Préparer les ingrédients (*les données / variables*)
2. Réaliser des étapes dans l'ordre (*les instructions*)
3. Obtenir un résultat (*l'affichage, un fichier…*)

```python
# La « recette » d'un programme très simple
nom = input("Comment t'appelles-tu ? ")   # 1. ingrédient
print("Bonjour", nom, "!")                   # 2. étape → 3. résultat
```

Comme en cuisine, **l'ordre compte** : on ne peut pas afficher `nom` avant de l'avoir récupéré.

````python
nom = "Ada"
print("Bonjour", nom, "!")
print("Bienvenue dans le monde de la programmation.")
````

# Les ingrédients d'un programme

Quels que soient le langage et le problème, presque tous les programmes reposent sur quelques notions clés :

| Notion           | Rôle                                          | Exemple Python            |
|------------------|-----------------------------------------------|---------------------------|
| **Variable**     | Stocker une valeur                            | `age = 25`                |
| **Opération**    | Calculer / transformer                        | `age + 1`                 |
| **Condition**    | Choisir ce qu'on fait selon une situation     | `if age >= 18: ...`       |
| **Boucle**       | Répéter une action plusieurs fois             | `for i in range(3): ...`  |
| **Fonction**     | Réutiliser un bloc de code                    | `print(...)`, `len(...)`   |
| **Entrée/Sortie**| Lire des données / afficher un résultat        | `input()`, `print()`      |

On voit toutes ces notions dans les prochains chapitres. Pour l'instant, retenez juste les **mots** : ils reviendront partout.

# L'ordinateur est littéral — et bête

L'ordinateur ne fait **aucune supposition**. Il ne « comprend » pas vos intentions : il suit vos instructions à la lettre.

```python
print("3" + "5")   # ???
```

Pour un humain, `3 + 5 = 8`. Pour Python, `"3"` et `"5"` sont du *texte*, donc il les *colle* : `"35"`. Ce n'est pas une erreur de Python — c'est vous qui lui avez menti sur le type des données.

> **Leçon n°1** : un programme fait exactement ce que vous lui demandez, pas ce que vous *pensiez* lui demander.

````python
print("3" + "5")   # collage de chaînes
print(3 + 5)     # vraie somme entière
````

# Conseil 1 — Pensez naïf, optimisez ensuite

Quand vous abordez un problème :
1. **Écrivez d'abord la solution la plus simple** qui fonctionne, même si elle vous paraît « lente » ou « maladroite ».
2. **Vérifiez qu'elle donne le bon résultat**.
3. **Seulement ensuite**, demandez-vous si elle est trop lente, trop longue, trop répétitive — et améliorez-la si besoin.

> 🧭 *« Premature optimization is the root of all evil. »* — Donald Knuth

Un code correct et lent sera toujours plus utile qu'un code censé être rapide mais faux ou jamais terminé.

```python
# Exemple : compter de 1 à 5
# Version naïve (parfaite pour commencer)
print(1)
print(2)
print(3)
print(4)
print(5)
```
Plus tard, on apprendra les boucles et on écrira `for i in range(1, 6): print(i)`. **Mais inutile de viser ça dès le premier jour !**

````python
# Version naïve — totalement acceptable quand on débute
print(1)
print(2)
print(3)
print(4)
print(5)
````

````python
# Version plus « élégante » — on la verra plus tard avec les boucles
for i in range(1, 6):
    print(i)
````

# Conseil 2 — Décomposez le problème

Face à un exercice qui semble trop grand, **coupez-le en morceaux**.

Exemple : *« Demander l'âge de l'utilisateur et afficher s'il est majeur. »*
Étape par étape :
1. Lire une saisie au clavier → `input()`
2. Convertir ce texte en nombre entier → `int()`
3. Comparer ce nombre à 18 → `>=`
4. Afficher un message différent selon le cas → `if / else`
5. Assembler le tout

> 🧩 Aucun problème n'est simple ou difficile *en soi* : il est juste plus ou moins **décomposable**. Apprenez à *découper*.

````python
# Décomposition appliquée
reponse = input("Quel est votre âge ? ")         # 1. lecture
age = int(reponse)                              # 2. conversion
est_majeur = age >= 18                          # 3. comparaison
if est_majeur:                                  # 4. choix
    print("Vous êtes majeur.")
else:
    print("Vous êtes mineur.")
````

# Conseil 3 — Lisez les messages d'erreur

Les erreurs ne sont pas des punitions : ce sont des **indices**. Python vous dit précisément *où* et *pourquoi* il s'est arrêté.

```python
>>> print("Hello)
  File "<stdin>", line 1
    print("Hello)
          ^
SyntaxError: unterminated string literal (detected at line 2)
```

Décryptage :
- **`SyntaxError`** : problème de syntaxe, le code est mal écrit.
- **`EOL while scanning string literal`** : fin de ligne atteinte alors qu'une chaîne n'était pas refermée → il manque un `"`.
- La flèche `^` pointe **exactement** l'endroit du problème.

> 🔍 Lisez **toujours** l'erreur, du bas vers le haut : la dernière ligne donne le type, les lignes au-dessus indiquent où.

# Conseil 5 — La lisibilité d'abord

Vous relirez votre code bien plus souvent que vous ne l'écrirez. Favorisez :

- des **noms de variables explicites** : `nombre_etudiants` plutôt que `n`, ou `prix_total` plutôt que `x`,
- un **espacement régulier** (convention PEP 8 : autour des `=`, des opérateurs),
- des **commentaires** quand le *pourquoi* n'est pas évident — pas pour dire *ce que fait* la ligne (le code le dit déjà).
- une **fonctionnalité à la fois** : un programme doit faire une chose claire, pas dix.

```python
# Bien
nombre_etudiants = 25
prix_total = prix_unitaire * nombre_etudiants

# Mal (code « golf »)
n = 25
pt = pu * n
```


# Conseil 6 — Cherchez avant de demander

Quand vous bloquez, suivez cet ordre intelligent :

1. **Relisez votre code** lentement, ligne par ligne.

4. **Cherchez sur Google / Stack Overflow** avec le message d'erreur exact.
5. **Demandez à un pair** ou au formateur — avec le contexte et ce que vous avez essayé.

> 🔎 Chercher une solution fait partie intégrante du métier. Personne, même expert, ne connaît tout par cœur.
