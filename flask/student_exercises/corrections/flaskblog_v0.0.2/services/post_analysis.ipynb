```python
from datetime import datetime

import pandas as pd
import plotly.express as px

from post import PostService, Post
```

```python
post_service = PostService()
```

```python
start_date: datetime = datetime(2026, 8, 11)
end_date: datetime = datetime(2026, 8, 15)

posts: list[Post] = post_service.get_by_dates(start_date, end_date)

df = pd.DataFrame([post.to_dict() for post in posts])
df["created_at"] = pd.to_datetime(df["created_at"]).dt.date
df["updated_at"] = pd.to_datetime(df["updated_at"]).dt.date
df
```
